```python
import cv2
import h5py
import numpy as np

video_path = r"C:\Users\jaycelehman\Desktop\extract_test\SB01\07242026_A\scope-2026-07-24T15_31_26.avi"
output_path = r"C:\Users\jaycelehman\Desktop\extract_test\SB01\07242026_A\scope-out.h5"

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise IOError("Cannot open video file")

frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

print(f"Frames: {frame_count}")
print(f"Size: {height} x {width}")

with h5py.File(output_path, "w") as h5f:
    dset = h5f.create_dataset(
        "data",
        shape=(frame_count, height, width),
        dtype="float32",
        chunks=(1, height, width),
    )

    frame_idx = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame.ndim == 3:
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        dset[frame_idx] = frame.astype(np.float32)
        frame_idx += 1

cap.release()

print("Done.")
```
